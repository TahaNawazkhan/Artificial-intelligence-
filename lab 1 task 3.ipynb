```python
import random

ACTIONS = ["Left", "Right", "Suck"]


def reflex_agent(percept):
    """Simple reflex agent: clean if dirty, otherwise move to the other square."""
    location, status = percept
    if status == "Dirty":
        return "Suck"
    return "Right" if location == "A" else "Left"


def random_agent(percept):
    """Random agent: ignores the percept and picks any action at random."""
    return random.choice(ACTIONS)


def random_initial_state():
    return {
        "loc": random.choice(["A", "B"]),
        "dirt": {"A": random.choice(["Clean", "Dirty"]),
                 "B": random.choice(["Clean", "Dirty"])},
    }


def apply_action(state, action):
    if action == "Suck":
        state["dirt"][state["loc"]] = "Clean"
    elif action == "Left":
        state["loc"] = "A"
    elif action == "Right":
        state["loc"] = "B"


def run(agent, initial_state, steps=20):
    """Performance = 1 point for every clean square at the end of each time step."""
    state = {"loc": initial_state["loc"], "dirt": dict(initial_state["dirt"])}
    score = 0
    for _ in range(steps):
        percept = (state["loc"], state["dirt"][state["loc"]])
        apply_action(state, agent(percept))
        score += sum(1 for s in state["dirt"].values() if s == "Clean")
    return score


if __name__ == "__main__":
    random.seed(42)                      # remove for different results each run
    STEPS, TRIALS = 20, 10
    initial_states = [random_initial_state() for _ in range(TRIALS)]

    print(f"{'Trial':<6}{'Start (loc, A, B)':<28}{'Reflex':>8}{'Random':>8}")
    reflex_scores, random_scores = [], []
    for i, init in enumerate(initial_states, 1):
        r = run(reflex_agent, init, STEPS)
        n = run(random_agent, init, STEPS)
        reflex_scores.append(r)
        random_scores.append(n)
        start = f"{init['loc']}, {init['dirt']['A']}, {init['dirt']['B']}"
        print(f"{i:<6}{start:<28}{r:>8}{n:>8}")

    print(f"\nAverage reflex agent score: {sum(reflex_scores) / TRIALS:.1f} / {2 * STEPS}")
    print(f"Average random agent score: {sum(random_scores) / TRIALS:.1f} / {2 * STEPS}")
```
