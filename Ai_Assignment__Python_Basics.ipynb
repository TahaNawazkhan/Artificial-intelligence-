```python

def section(title):
    print("\n" + "=" * 50)
    print(title)
    print("=" * 50)


# 1. Variables and data types
section("1. VARIABLES & DATA TYPES")
age = 25                  # int
price = 19.99             # float
name = "Ali"              # str
is_student = True         # bool
nothing = None            # NoneType
complex_num = 3 + 4j      # complex

print("int:", age, type(age))
print("float:", price, type(price))
print("str:", name, type(name))
print("bool:", is_student, type(is_student))
print("None:", nothing, type(nothing))
print("complex:", complex_num, type(complex_num))

# Type conversion
print("int('42') + 8 =", int("42") + 8)
print("float(7) =", float(7))
print("str(100) + '!' =", str(100) + "!")

# 2. Operators
section("2. OPERATORS")
a, b = 17, 5
print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)       # true division
print("a // b =", a // b)     # floor division
print("a % b =", a % b)       # remainder
print("a ** b =", a ** b)     # power
print("a > b and b > 0:", a > b and b > 0)
print("a < b or b == 5:", a < b or b == 5)
print("not (a == b):", not (a == b))

# 3. Strings
section("3. STRINGS")
text = "Hello, Python!"
print("Upper:", text.upper())
print("Lower:", text.lower())
print("Length:", len(text))
print("Slice [0:5]:", text[0:5])
print("Reversed:", text[::-1])
print("Replace:", text.replace("Python", "World"))
print(f"f-string: {name} is {age} years old and price is {price:.1f}")

# 4. If / elif / else
section("4. IF / ELIF / ELSE")
marks = 78
if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"
print(f"Marks: {marks} -> Grade: {grade}")

# Ternary (one-line if/else)
status = "Pass" if marks >= 50 else "Fail"
print("Status:", status)

# 5. Switch (match / case, Python 3.10+)
section("5. SWITCH (match / case)")


def day_name(num):
    match num:
        case 1:
            return "Monday"
        case 2:
            return "Tuesday"
        case 3:
            return "Wednesday"
        case 4 | 5:
            return "Thursday or Friday"
        case 6 | 7:
            return "Weekend"
        case _:
            return "Invalid day"


for n in (1, 3, 5, 7, 9):
    print(f"Day {n}: {day_name(n)}")

# 6. Loops
section("6. FOR LOOP")
for i in range(1, 6):
    print("i =", i)

print("Even numbers 2..10:", [x for x in range(2, 11, 2)])

section("7. WHILE LOOP")
count = 1
while count <= 5:
    print("count =", count)
    count += 1

section("8. DO-WHILE (Python has none; use while True + break)")
num = 10
while True:
    print("This runs at least once, num =", num)
    num += 1
    if num > 12:
        break

section("9. BREAK, CONTINUE, PASS, LOOP ELSE")
for i in range(1, 10):
    if i == 3:
        continue          # skip 3
    if i == 7:
        break             # stop at 7
    print("loop value:", i)
else:
    print("This else runs only if loop was not broken")

for i in range(3):
    pass                  # placeholder, does nothing

# 10. Lists (arrays)
section("10. LISTS (ARRAYS)")
fruits = ["apple", "banana", "cherry"]
fruits.append("mango")
fruits.insert(1, "orange")
fruits.remove("banana")
print("List:", fruits)
print("First:", fruits[0], "| Last:", fruits[-1])
print("Length:", len(fruits))
print("Sorted:", sorted(fruits))

numbers = [5, 2, 9, 1, 7]
print("Numbers:", numbers)
print("Sum:", sum(numbers), "| Max:", max(numbers), "| Min:", min(numbers))
numbers.sort()
print("Sorted numbers:", numbers)

# 2D array
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("Matrix row 1:", matrix[1], "| element [2][0]:", matrix[2][0])
for row in matrix:
    print(row)

# 11. Tuples, Sets, Dictionaries
section("11. TUPLE, SET, DICTIONARY")
point = (10, 20)                  # immutable
x, y = point
print("Tuple:", point, "| unpacked x, y =", x, y)

unique = {1, 2, 2, 3, 3, 3}       # no duplicates
unique.add(4)
print("Set:", unique)

student = {"name": "Sara", "age": 21, "course": "CS"}
student["gpa"] = 3.8
print("Dict:", student)
for key, value in student.items():
    print(f"  {key}: {value}")

# 12. Functions
section("12. FUNCTIONS")


def add(a, b=0):
    """Add two numbers (b is optional)."""
    return a + b


def factorial(n):
    return 1 if n <= 1 else n * factorial(n - 1)   # recursion


def stats(*nums):
    """Variable arguments, returns multiple values."""
    return min(nums), max(nums), sum(nums) / len(nums)


print("add(3, 4) =", add(3, 4))
print("add(5) =", add(5))
print("factorial(5) =", factorial(5))
lo, hi, avg = stats(4, 8, 15, 16, 23, 42)
print(f"stats -> min={lo}, max={hi}, avg={avg:.2f}")

square = lambda n: n * n
print("lambda square(6) =", square(6))
print("map:", list(map(square, [1, 2, 3, 4])))
print("filter evens:", list(filter(lambda n: n % 2 == 0, range(1, 11))))

# 13. Exception handling
section("13. EXCEPTION HANDLING")
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Error: cannot divide by zero")
except Exception as e:
    print("Other error:", e)
else:
    print("No error, result =", result)
finally:
    print("finally always runs")

try:
    int("abc")
except ValueError as e:
    print("ValueError caught:", e)

# 14. Classes and objects
section("14. CLASSES & OBJECTS (OOP)")


class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.sound = sound

    def speak(self):
        return f"{self.name} says {self.sound}"


class Dog(Animal):                 # inheritance
    def __init__(self, name):
        super().__init__(name, "Woof")

    def fetch(self):
        return f"{self.name} fetches the ball"


cat = Animal("Cat", "Meow")
dog = Dog("Buddy")
print(cat.speak())
print(dog.speak())
print(dog.fetch())

# 15. File handling
section("15. FILE HANDLING")
with open("demo.txt", "w") as f:
    f.write("Line 1\nLine 2\nLine 3\n")

with open("demo.txt", "r") as f:
    for line_no, line in enumerate(f, start=1):
        print(f"{line_no}: {line.strip()}")

# 16. Input (safe demo: only asks if run in a terminal)
section("16. USER INPUT")
try:
    user = input("Enter your name (or press Enter to skip): ").strip()
    print(f"Hello, {user or 'Guest'}!")
except EOFError:
    print("No input available, skipping.")

# 17. Mini program: simple menu calculator using everything above
section("17. MINI PROGRAM - CALCULATOR")


def calculate(op, x, y):
    match op:
        case "+":
            return x + y
        case "-":
            return x - y
        case "*":
            return x * y
        case "/":
            return x / y if y != 0 else "Error: division by zero"
        case _:
            return "Unknown operation"


for op in ["+", "-", "*", "/", "%"]:
    print(f"12 {op} 4 =", calculate(op, 12, 4))

print("\nDone! You covered all the Python basics.")
```
