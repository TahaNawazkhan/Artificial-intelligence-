```python
MAZE = [
    "S.#....",
    "#.#.##.",
    "....#..",
    ".####.#",
    ".......",
    "######.",
    "......G",
]
ROWS, COLS = len(MAZE), len(MAZE[0])


def locate(ch):
    for r in range(ROWS):
        if ch in MAZE[r]:
            return (r, MAZE[r].index(ch))


def children(cell):
    """Open neighbouring cells (Up, Down, Left, Right)."""
    r, c = cell
    for dr, dc in ((-1, 0), (1, 0), (0, -1), (0, 1)):
        nr, nc = r + dr, c + dc
        if 0 <= nr < ROWS and 0 <= nc < COLS and MAZE[nr][nc] != "#":
            yield (nr, nc)


def dls(cell, goal, limit, visited_order, path):
    """Recursive Depth-Limited Search.
    Returns 'found', 'cutoff' or 'failure'.
    `path` holds the cells on the current branch (stops the search from
    walking in circles, which a maze - unlike a tree - allows)."""
    visited_order.append(cell)
    if cell == goal:
        return 'found'
    if limit == 0:
        return 'cutoff'
    cutoff_occurred = False
    for child in children(cell):
        if child in path:
            continue
        path.append(child)
        res = dls(child, goal, limit - 1, visited_order, path)
        if res == 'found':
            return 'found'
        path.pop()
        if res == 'cutoff':
            cutoff_occurred = True
    return 'cutoff' if cutoff_occurred else 'failure'


start, goal = locate('S'), locate('G')

print("Maze:")
for row in MAZE:
    print("  ", " ".join(row))
print(f"\nStart = {start}, Goal = {goal}\n")

print(f"{'Limit':>5} | {'Result':<8} | {'Nodes visited':>13}")
print("-" * 33)
for limit in (5, 10, 15):
    order, path = [], [start]
    result = dls(start, goal, limit, order, path)
    print(f"{limit:>5} | {result:<8} | {len(order):>13}")
    if result == 'found':
        print(f"        path ({len(path) - 1} steps): "
              + " -> ".join(f"({r},{c})" for r, c in path))
```
