```python
import heapq

MOVES = {"Up": (-1, 0), "Down": (1, 0), "Left": (0, -1), "Right": (0, 1)}


def manhattan(state, goal):
    pos = {v: i for i, v in enumerate(goal)}
    total = 0
    for i, v in enumerate(state):
        if v != 0:
            total += abs(i // 3 - pos[v] // 3) + abs(i % 3 - pos[v] % 3)
    return total


def apply_move(state, move):
    """Move the blank (0) in the given direction and return the new state."""
    b = state.index(0)
    r, c = divmod(b, 3)
    dr, dc = MOVES[move]
    n = (r + dr) * 3 + (c + dc)
    s = list(state)
    s[b], s[n] = s[n], s[b]
    return tuple(s)


def format_board(state):
    rows = []
    for r in range(3):
        rows.append(" ".join(" " if v == 0 else str(v) for v in state[3 * r:3 * r + 3]))
    return rows


def print_solution(start, path):
    """Print the board after every move. `path` is a list like ['Up', 'Left', ...]."""
    state = start
    print("Start")
    print("\n".join(format_board(state)))
    for step, move in enumerate(path, 1):
        state = apply_move(state, move)
        print(f"\nStep {step}: move blank {move}")
        print("\n".join(format_board(state)))
    print(f"\nSolved in {len(path)} moves.")


def solve(start, goal):
    """A* search with Manhattan distance. Returns the list of moves from start to goal."""
    frontier = [(manhattan(start, goal), 0, 0, start)]
    parent = {start: None}          # state -> (previous state, move)
    best_g = {start: 0}
    tie = 0
    while frontier:
        f, g, _, state = heapq.heappop(frontier)
        if g > best_g[state]:
            continue
        if state == goal:
            path = []
            while parent[state] is not None:
                state, move = parent[state]
                path.append(move)
            return path[::-1]
        b = state.index(0)
        r, c = divmod(b, 3)
        for move, (dr, dc) in MOVES.items():
            if 0 <= r + dr < 3 and 0 <= c + dc < 3:
                nxt = apply_move(state, move)
                ng = g + 1
                if ng < best_g.get(nxt, float("inf")):
                    best_g[nxt] = ng
                    parent[nxt] = (state, move)
                    tie += 1
                    heapq.heappush(frontier, (ng + manhattan(nxt, goal), ng, tie, nxt))
    return None


if __name__ == "__main__":
    start = (8, 1, 3, 4, 0, 2, 7, 6, 5)
    goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)
    path = solve(start, goal)
    print("Path:", path, "\n")
    print_solution(start, path)
```
